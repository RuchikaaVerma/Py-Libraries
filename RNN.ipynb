RNN: Recurrent Neural Network are the first of its State of the art algo that can memorize/remember previous inputs in memory.When a huge set of sequentail data is given . First kind of state of art that can  memozize when huge set of sequentail data is given.

These loops make recurrent neural network seem  kind of mysterious .A recurrent neural network is mutiple copies of same network each passing a message to a successor.
Different types of recurrent relations:-
-Image classification (single input and single output eg:image captioning takes an image and iutputs a sentence of words.)
-Sequence output(eg: sentiment analysis where a given sentence is classified as expressing postive or negative sentiments.)
-sequence input and sequence output(eg:Machine Translation:an RNN reads a sentence in english and then output a sentence in french.)
-Syed sequence input and output(eg:video classsification where we wish to label frame to the vedio.)

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
#types
def draw_box(ax, x, y, color='lightgrey', label=None):
    rect = patches.Rectangle((x, y), 0.6, 1, linewidth=1, edgecolor='black', facecolor=color)
    ax.add_patch(rect)
    if label:
        ax.text(x + 0.3, y + 0.5, label, ha='center', va='center', fontsize=10)

def draw_arrow(ax, start, end):
    ax.annotate('', xy=end, xytext=start,
                arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5))

def plot_rnn_types():
    fig, axs = plt.subplots(1, 5, figsize=(22, 5))
    plt.subplots_adjust(wspace=0.6)

    titles = ['one to one', 'one to many', 'many to one', 'many to many', 'many to many (seq)']

    for ax, title in zip(axs, titles):
        ax.set_title(title, fontsize=14)
        ax.axis('off')
        ax.set_xlim(0, 4)
        ax.set_ylim(0, 5)

    # 1) one to one
    ax = axs[0]
    draw_box(ax, 1.7, 1, 'pink', 'Input')
    draw_box(ax, 1.7, 2.5, 'lightgrey')
    draw_box(ax, 1.7, 4, 'skyblue', 'Output')

    draw_arrow(ax, (2, 1 + 1), (2, 2.5))
    draw_arrow(ax, (2, 2.5 + 1), (2, 4))

    # 2) one to many
    ax = axs[1]
    draw_box(ax, 1.7, 1, 'pink', 'Input')

    # hidden states
    draw_box(ax, 0.5, 2.5, 'lightgrey')
    draw_box(ax, 1.7, 2.5, 'lightgrey')
    draw_box(ax, 2.9, 2.5, 'lightgrey')

    # outputs
    draw_box(ax, 0.5, 4, 'skyblue')
    draw_box(ax, 1.7, 4, 'skyblue')
    draw_box(ax, 2.9, 4, 'skyblue')

    # arrows input to hidden
    draw_arrow(ax, (2, 1 + 1), (2, 2.5))
    # hidden to output
    draw_arrow(ax, (0.8, 2.5 + 1), (0.8, 4))
    draw_arrow(ax, (1.9, 2.5 + 1), (1.9, 4))
    draw_arrow(ax, (3.1, 2.5 + 1), (3.1, 4))

    # horizontal arrows between hidden states
    draw_arrow(ax, (1.1, 3), (1.5, 3))
    draw_arrow(ax, (2.3, 3), (2.7, 3))

    # 3) many to one
    ax = axs[2]
    # inputs
    draw_box(ax, 0.5, 1, 'pink')
    draw_box(ax, 1.7, 1, 'pink')
    draw_box(ax, 2.9, 1, 'pink')

    # hidden states
    draw_box(ax, 0.5, 2.5, 'lightgrey')
    draw_box(ax, 1.7, 2.5, 'lightgrey')
    draw_box(ax, 2.9, 2.5, 'lightgrey')

    # output
    draw_box(ax, 1.7, 4, 'skyblue', 'Output')

    # arrows input to hidden
    draw_arrow(ax, (0.8, 1 + 1), (0.8, 2.5))
    draw_arrow(ax, (1.9, 1 + 1), (1.9, 2.5))
    draw_arrow(ax, (3.1, 1 + 1), (3.1, 2.5))

    # horizontal arrows between hidden states
    draw_arrow(ax, (1.1, 3), (1.5, 3))
    draw_arrow(ax, (2.3, 3), (2.7, 3))

    # arrows hidden to output
    draw_arrow(ax, (1.9, 2.5 + 1), (1.9, 4))

    # 4) many to many (sequence)
    ax = axs[3]
    xs = [0.5, 1.7, 2.9]
    for x in xs:
        draw_box(ax, x, 1, 'pink')
        draw_box(ax, x, 2.5, 'lightgrey')
        draw_box(ax, x, 4, 'skyblue')

    # vertical arrows input -> hidden -> output
    for x in xs:
        draw_arrow(ax, (x + 0.3, 1 + 1), (x + 0.3, 2.5))
        draw_arrow(ax, (x + 0.3, 2.5 + 1), (x + 0.3, 4))

    # horizontal arrows between hidden states
    draw_arrow(ax, (0.8, 3), (1.4, 3))
    draw_arrow(ax, (2.0, 3), (2.6, 3))

    # 5) many to many (sequence full)
    ax = axs[4]
    xs = [0.5, 1.7, 2.9]
    for x in xs:
        draw_box(ax, x, 1, 'pink')
        draw_box(ax, x, 2.5, 'lightgrey')
        draw_box(ax, x, 4, 'skyblue')

    for x in xs:
        draw_arrow(ax, (x + 0.3, 1 + 1), (x + 0.3, 2.5))
        draw_arrow(ax, (x + 0.3, 2.5 + 1), (x + 0.3, 4))

    # horizontal arrows between hidden states
    draw_arrow(ax, (0.8, 3), (1.4, 3))
    draw_arrow(ax, (2.0, 3), (2.6, 3))

    # horizontal arrows between inputs
    draw_arrow(ax, (0.8, 1.5), (1.4, 1.5))
    draw_arrow(ax, (2.0, 1.5), (2.6, 1.5))

    plt.show()

plot_rnn_types()
```

The problem of RNN or long term dependencies:

-vanishing Gradient
-Exploding Gradient

Vanishing Gradient:
If the partial derivation of error is less than 1,then when it get multiplied then the learning rate which is also very less,then multipyling leanrning  rate with partial derivation of error wont be a big change when compared with previous alteration.



Long Short Term  Memory (LSTM) Networks:
A special kind of RNN ,capable of leaning long term dependencies.Explicit designed to avoid the long term dependency problem.remembering info for long periods of time .

Building LSTM Model and Prepare X and Y 

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
token= Tokenizer()
token.fit_on_texts(data)
token.word_counts

```

```python
help(token)
```

```python
token.word_index
```

```python
encoded_text=token.texts_to_sequences(data)
encoded_text
```

```python
vocab_size=len(token.word_counts) +1
```

Prepare Training Data

```python
datalist=[]
for d in encoded_text:
    if len(d)>1:
        for i in range(2,len(d)):
            datalist.append(d[:i])
    print(d[:i])
```

Padding

```python
max_length=20
sequences=pad_sequences(datalist,maxlen=max_length ,padding='pre')
sequences
```

```python
x=sequences[:, :-1]
```

```python
y=sequences[:,-1]
```

```python
vocab_size
```

```python
y=to_categorical(y,num_classes=vocab_size)
y
```

```python
seq_length=x.shape[1]
seq_length
```

LSTM Model Training


```python
from tensorflow.keras.layers import Embedding, LSTM
```

```python
model =Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=100))
model.add(LSTM(100,return_sequences=True))
model.add(LSTM(100))
model.add(Dense(100,activation='relu'))
model.add(Dense(vocab_size, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
print(type(x), len(x))
print(type(y), len(y))
```

```python
print(x.shape)
print(y.shape)
```

```python
model.fit(x,y, batch_size=32, epochs=50)

```

Poetry Generation

```python
poetry_length=10
def generate_poetry(seed_text,n_lines):
    for i in range(n_lines):
        text=[]
        for _ in range(poetry_length):
            encoded=token.texts_to_sequences([seed_text])
            encoded=pad_sequences(encoded,maxlen=seq_length ,padding='pre')
            
            y_pred=np.argmax(model.predict(encoded),axis=1)
            
            predicted_word=""
            for word,index in token.word_index.items():
                if index ==y_pred:
                    predicted_word=word
                    break
                
            seed_text=seed_text +'' +predicted_word
            text.append(predicted_word) 
        seed_text=text[-1]    
        text='' .join(text)
        print(text)
```
