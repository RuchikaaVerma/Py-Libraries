```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()
model=ChatOpenAI(model="gpt-4",temperature=0.9)#0-2=creativity and determinism response
result=model.invoke("What is the capital of France?")
print(result)
```

```python
#claude -Anthropic's language model
from lanchain_anthropic import ChatAnthropic
from dotenv import load_dotenv

load_dotenv()
model=ChatAnthropic(model="claude-2",temperature=0.3)
result=model.invoke("What is the capital of France?")
print(result)
```

```python
#Google_gemini
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
load_dotenv()
model=ChatGoogleGenerativeAI(model="gemini-1.5-pro",temperature=0.3)
result=model.invoke("What is the capital of France?")
print(result)

```
