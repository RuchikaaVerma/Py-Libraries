```python
!pip install langchain-openai
from langchain_openai import OpenAI
from dotenv import load_dotenv#load secret keys

#invoke the load_dotenv function to load the environment variables from the .env file
load_dotenv()
#object of the OpenAI class

llm = OpenAI(model="gpt-3.5-turbo-instruct")
#call the llm object with a prompt to generate a response
result=llm.invoke("What is the capital of France?")
print(result)
```
