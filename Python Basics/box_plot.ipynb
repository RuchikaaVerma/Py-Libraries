#Minimum,MAximum,medain,Q1,Q3,IQR

```python
import numpy as np
```

```python
lst_marks=[23,45,67,89,12,34,56,78,90,11,22,33,44,55,66,77,88,99,100]
minimum,Q1,median,Q3,maximum=np.quantile(lst_marks,[0,0.25,0.5,0.75,1])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
#outliers IQR
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*(IQR)
higher_fence=Q3+1.5*(IQR)
#gretaer nd lower than number are outliers
```

```python
lower_fence
```

```python
higher_fence
```
