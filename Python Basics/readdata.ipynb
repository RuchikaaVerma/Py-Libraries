```python
import pandas as pd
from io import StringIO

# Properly formatted JSON string as a list of objects
Data = '[{"empolyee_name":"James","empolyee_id":1234,"department":"HR"},' \
       '{"empolyee_name":"Michael","empolyee_id":5678,"department":"Finance"},' \
       '{"empolyee_name":"Robert","empolyee_id":9101,"department":"IT"},' \
       '{"empolyee_name":"Maria","empolyee_id":1121,"department":"Admin"}]'

# Read JSON into DataFrame
df = pd.read_json(StringIO(Data))

print(df)
```
