Reference counting is the primary method python uses to manage memory.Each object in Python maintains a count referneces pointing to it.When the reference count dropsto zero,the memory occupied by the object is deallocated.


```python
import sys 
a=[]
print(sys.getrefcount(a)) # 2(one from 'a' and one from getrefcount argument)
```

```python
b=a
print(sys.getrefcount(a)) # 3(one from 'a',one from 'b
```

```python
del b
print(sys.getrefcount(a)) 
```

Garbage collection

Python uses a cyclic garbage collector to detect and clean up reference cycles,which are groups of
objects that reference each other but are no longer reachable from the program. The garbage collector periodically looks for these cycles and frees the memory they occupy.

```python
import gc
gc.enable()
# Enable automatic garbage collection
```

```python
gc.disable()
```

```python
gc.collect()#unreachable object are showed
```

```python
print(gc.get_stats())
```

```python
#get unreachable objects
print(gc.garbage)
```

Best Practices

1.LocalVariable: shorter kifespan and are freed sooner than global variables

2.Avoid Circular Referneces:Lead to memory leaks if not properly managed.

3.Use generators:Generators produce items one at a time and only keep item in memory at a time making them memory efficient.

4.Explicitly delete Objects:Use the del statement to delete variables and objects explicitly.

5.Memory Usage: Use memory profiling tools like tracemalloc and memory_profilier to identify memory leaks and optimize memory usage .

```python
#handling circular refernces
import gc
class MyObject:
    def __init__(self, name):
        self.name = name
        print(f"{self.name} created")
    def __del__(self):
        print(f"{self.name} destroyed")
        
#create circular reference
obj1=MyObject("Object 1")
obj2=MyObject("Object 2")
obj1.ref=obj2
obj2.ref=obj1

del obj1
del obj2
#manually trigger gc
gc.collect()
```

```python
#Generatorsfor Memory efficiency:allow you to produce items one at a time ,using memory efficiently only by keeping one item in memory at a time.
def generate_numbers(n):
    for i in range(n):
        yield i
#using generator
for number in generate_numbers(10):
    print(number)
    if number>=5:
        break
```

```python
#Profillingmemeory usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()
    create_list()
    
    snapshot= tracemalloc.take_snapshot()
    tap_stats=snapshot.statistics('lineno')

    print("[ Top 10 memory consuming lines ]")
    for stat in tap_stats[:10]:
        print(stat)
```

```python
main()
```
