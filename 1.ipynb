```python
import matplotlib.pyplot as plt

# Test cases (keys searched)
keys = [10, 25, 45, 75, 50, 100]
# Expected index results (-1 means not found)
results = [0, 1, 3, 5, 4, -1]
# Simulated execution times (small differences)
execution_times = [0.000001, 0.0000012, 0.0000018, 0.000002, 0.0000019, 0.0000023]

plt.figure(figsize=(9,6))

# Plot line graph
plt.plot(keys, execution_times, marker="o", linestyle="-", color="blue", label="Execution Time")

# Annotate each point
for i, (k, t) in enumerate(zip(keys, execution_times)):
    if results[i] != -1:
        plt.text(k, t+0.0000001, f"idx={results[i]}", ha="center", color="green", fontweight="bold")
    else:
        plt.text(k, t+0.0000001, "Not Found", ha="center", color="red", fontweight="bold")

# Labels and title
plt.xlabel("Key Searched")
plt.ylabel("Execution Time (seconds)")
plt.title("Recursive Linear Search: Execution Time vs Key")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Recursive Binary Search implementation
def recursive_binary_search(arr, target, low=0, high=None):
    if high is None:
        high = len(arr) - 1
    if low > high:
        return -1  # not found
    mid = (low + high) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return recursive_binary_search(arr, target, low, mid - 1)
    else:
        return recursive_binary_search(arr, target, mid + 1, high)

# Test cases (keys searched)
keys = [10, 25, 45, 75, 50, 100]
# Sorted array for binary search context
arr = sorted([10, 25, 35, 45, 50, 75])
# Expected index results (-1 means not found)
results = [recursive_binary_search(arr, k) for k in keys]
# Simulated execution times (binary search is faster)
execution_times = [0.0000006, 0.0000008, 0.0000011, 0.0000013, 0.0000010, 0.0000015]

plt.figure(figsize=(9,6))

# Plot line graph
plt.plot(keys, execution_times, marker="o", linestyle="-", color="blue", label="Execution Time")

# Annotate each point
for i, (k, t) in enumerate(zip(keys, execution_times)):
    if results[i] != -1:
        plt.text(k, t+0.00000005, f"idx={results[i]}", ha="center", color="green", fontweight="bold")
    else:
        plt.text(k, t+0.00000005, "Not Found", ha="center", color="red", fontweight="bold")

# Labels and title
plt.xlabel("Key Searched")
plt.ylabel("Execution Time (seconds)")
plt.title("Recursive Binary Search: Execution Time vs Key")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```
