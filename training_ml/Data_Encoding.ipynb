Nominal/OHE Encoding

One hot encoding ,also known as nominal encoding,is a technique used to represent categorialdata as numerical data,which is more suitable for ML algorithms.each category is represented as a binary vector where each bit corresponds to a unique category.

For example-
Having 3 colors(Red , BLue,Green):{categorial feature to numerical feature}

Red(1,0,0),blue(0,1,0),Green(0,0,1)

many categorial datset not uses this it give overfitting error



```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame({
    'color':['red','blue','green','yellow','purple']
})
```

```python
df.head()
```

```python
#create an instance oh OHE
encoder=OneHotEncoder()
```

```python
#perform fit and transform
encoded=encoder.fit_transform(df[['color']]).toarray()#sprase matrix
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
#for new data
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

Label Encoding

Used to encode categorical data as numerical data

Involves assigning a unique numeric data label to each category in the cariable.The label are usually assigned in alphabetical order or based on the frequency of the categories.

Example:3-color(red,blure,green)-

red=1,blue=2,green=3

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder=LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

Ordinal Encoding

Used to encode categorical data that have a n intrinsic order or ranking. Each category is assigned a numerical value baes on its position in the order.For eg,categorical  variable"education level" with possible values(high school,colledge,graduate,post-graduate),we can represent it using ordinal encoding:

HS:1,C:2,G:3,PG:4

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    'size':['small','large','medium','small','large','medium']
})
```

```python
df
```

```python
#create an instance
encoder=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```

3. Target Guided Ordinal Encoding

Used to encode categorical variable on their relationship with the target value .The encoding technique is useful when we have a categorical variable with a large number of unique categories,we want to use this variable as faeture in our machine learning model.

We replace category in the categorical varibale with a numerical value based on the mean or median of the target varibale for that category.This creates a monotonic relationship between the categoriacl variables and he target variable,which can improve the predictive power of our model.

```python
import pandas as pd
df=pd.DataFrame({
    'city':['New York','London','Paris','tokyo','New Zealand'],
    'price':[200,300,400,230,400]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df
```

```python
df[['price','city_encoded']]
```
