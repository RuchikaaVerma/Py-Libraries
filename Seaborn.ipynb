Data visualoization with Seaborn
Seaborn is a python visualization library based on matplotlib that provides a high-level interafce for drawing interactive and informative statistical graphics. Seaborn hepls in creating complex visualization with just a few lines of code.


```python
import seaborn as sns
```

```python
#basic plotting with seaborn
tips=sns.load_dataset('tips')
print(tips.head())
```

```python
#scatter plot
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='day',style='time')
#show the plot
import matplotlib.pyplot as plt
plt.show()
```

```python
#line plot
sns.lineplot(data=tips,x='total_bill',y='tip',hue='day')
plt.show()
```

```python
#categorical plot
sns.catplot(data=tips,x='day',y='total_bill')
plt.show()
```

```python
#bar plot
sns.barplot(data=tips,x='day',y='total_bill',)
plt.show()
```

```python
#box plot
sns.boxplot(data=tips,x='day',y='total_bill',hue='smoker')
plt.show() 
```

```python
#violin plot
sns.violinplot(data=tips,x='day',y='total_bill',hue='smoker',split=True)
plt.show()
```

```python
##histogram
sns.histplot(data=tips,x='total_bill',bins=10,kde=True)##false:no curve
plt.show() 
```

```python
sns.histplot(data=tips,x='total_bill',bins=10,kde=False)##false:no curve
plt.show() 
```

```python
#kde plot
sns.kdeplot(data=tips,x='total_bill',hue='day',fill=True)
plt.show()
```

```python
#pair plot
sns.pairplot(tips,hue='day')
plt.show()
```

```python
flights = sns.load_dataset('flights')
flights = flights.pivot(index='month', columns='year', values='passengers')

# Display the pivoted DataFrame
print(flights)

# Plot the heatmap
sns.heatmap(flights, annot=True, fmt='.0f', cmap='YlGnBu')  # Use '.0f' for float formatting
plt.title("Monthly Airline Passengers (1949–1960)")
plt.show()
```

```python
corr=tips[['total_bill','tip','size']].corr()
print(corr)
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.show()
```
